```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, classification_report, confusion_matrix
```

```python
f_train = pd.read_csv('./data/train.csv')
f_test = pd.read_csv('./data/test.csv')
```

```python
import numpy as np

f_train = f_train.dropna()
f_test = f_test.dropna()

f_train['satisfactionn'] = np.where(f_train['satisfaction'] == 'satisfied', 1, 0)
f_test['satisfactionn'] = np.where(f_test['satisfaction'] == 'satisfied', 1, 0)

f_train.head(20)
```

```python

f_train.nunique()
```

```python

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier

X_train = f_train.drop(columns=['satisfaction', 'satisfactionn'])
y_train = f_train['satisfactionn']
X_test = f_test.drop(columns=['satisfaction', 'satisfactionn'])
y_test = f_test['satisfactionn']

numeric_features = X_train.select_dtypes(include=[np.number]).columns.tolist()
categorical_features = X_train.select_dtypes(include=['object']).columns.tolist()

numeric_transformer = Pipeline(steps=[
    ('scaler', StandardScaler())  # Standardisatie
])

categorical_transformer = Pipeline(steps=[
    ('onehot', OneHotEncoder(handle_unknown='ignore'))  # OH encoding voor categorische features
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ])

model = Pipeline([
    ('preprocessor', preprocessor),  # Preprocessing steps
    ('classifier', RandomForestClassifier(n_estimators=100, random_state=42))  # RandomForest Classifier
])
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)

# Evaluatie
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
class_report = classification_report(y_test, y_pred)

print(f"Accuracy: {accuracy:.2f}")
print("Confusion Matrix:")
print(conf_matrix)
print("Classification Report:")
print(class_report)
```
