```python
import numpy as np
import pandas as pd


df = pd.read_csv('winequality-red.csv')
df.head()
# Hier wil ik naar ruwe data zien
```

```python
df.describe()
# Dit geeft meer inzicht over wat ik precies heb
```

```python
# Ik kijk of preprocessing nodig is
print(df.isna().sum())
print(df.dtypes)
# Het lijkt dat ik moet niks imputeren
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# Preprocessing

X = df.drop(columns=['quality'])
y = df['quality']

numeric_features = X.columns
# Alle features zijn numeriek en continieu, daarom moet er geen encoding gedaan

numeric_transformer = Pipeline(steps=[
    ('scaler', StandardScaler())
])
# Hier zet ik alle numerieke waarden in dezelfde schaal

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features)
    ])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
# Deze dataset is klein maar niet te klijn, daarom geef ik aan de test_size default waarde van 0.25
```

```python
# Hier verwerk ik quality as labels

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

knn_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('knn', KNeighborsClassifier(n_neighbors=10))
])

# Eerst wil ik KNN testen, want die is geen te ingewikkeld model is en kan een goede baseline zijn om te zien of andere modelen kunnen beter presteren

knn_pipeline.fit(X_train, y_train)

# Predictions
y_pred = knn_pipeline.predict(X_test)

# Evaluatie
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
class_report = classification_report(y_test, y_pred)

print(f"Accuracy: {accuracy:.2f}")
print("Confusion Matrix:")
print(conf_matrix)
print("Classification Report:")
print(class_report)
```

```python
# Van deze resultaat kan ik al zien dat KNN iets kan bepalen maar 60% accuracy is nog niet te goed
```

```python
# Ik denk dat Random Forest gaat beter voor dit probleem zijn
from sklearn.ensemble import RandomForestClassifier

random_forest_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('rnd_forest', RandomForestClassifier(n_estimators=100, random_state=42))
])

random_forest_pipeline.fit(X_train, y_train)

# Predictions
y_pred = random_forest_pipeline.predict(X_test)

# Evaluatie
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
class_report = classification_report(y_test, y_pred)

print(f"Accuracy: {accuracy:.2f}")
print("Confusion Matrix:")
print(conf_matrix)
print("Classification Report:")
print(class_report)
```

```python
# Het is een improvement, f1-score ziet nu veel beter uit
```

```python
# Ten laatste wil ik de sterkste gebruiken - XGBoost
from xgboost import XGBClassifier

xgb_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('xgb_forest', XGBClassifier(n_estimators=50, random_state=42))
])

xgb_pipeline.fit(X_train, y_train)

# Predictions
y_pred = xgb_pipeline.predict(X_test)

# Evaluatie
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
class_report = classification_report(y_test, y_pred)

print(f"Accuracy: {accuracy:.2f}")
print("Confusion Matrix:")
print(conf_matrix)
print("Classification Report:")
print(class_report)

# Ik heb niet genoeg tijd om in eerste deel deze fout opzoeken, maar normal gezien XGB het beeste resultaten geeft, door zijn speciale gewichtverdelingen aan de trees
```

```python
# Hier verwerk ik 'quality' als regressie probleem

# Laat even SVM als regressor testen
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import GridSearchCV
svr_pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('svr', SVR(kernel="poly"))
])

parameters = {'svr__kernel': ['poly', 'linear'], 'svr__C':[0.01, 0.1, 5, 10]}

grid_svr = GridSearchCV(svr_pipeline, parameters)

grid_svr.fit(X_train, y_train)
# Hier experimenteer ik met grid search om de beste hyper parameters te vinden

# Predictions
y_pred = grid_svr.predict(X_test)
mse_svr = mean_squared_error(y_test, y_pred)
r2_svr = r2_score(y_test, y_pred)
print(mse_svr)
print(r2_svr)
print(grid_svr.best_params_)
```
